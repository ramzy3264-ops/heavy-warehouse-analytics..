# Heavy Warehouse Analytics: Phase 4 Visualizations & BI Insights
This notebook loads the processed analytical outputs from the `outputs/` directory and generates key visual summaries for inventory classification, customer RFM segmentation, and predictive reorder points.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Set plot styling
sns.set_theme(style="whitegrid")
plt.rcParams['figure.figsize'] = (10, 5)
```

## 1. Customer RFM Segmentation Distribution (Catalogue E)

```python
rfm_df = pd.read_csv('../outputs/customer_rfm_segments.csv')
print(rfm_df.head())

# Visualize Monetry vs Frequency
plt.figure(figsize=(8, 4))
sns.scatterplot(data=rfm_df, x='frequency', y='monetary', hue='R_Score', palette='viridis')
plt.title('Customer RFM Segmentation: Frequency vs Monetary Value')
plt.xlabel('Transaction Frequency')
plt.ylabel('Monetary Quantity')
plt.show()
```

## 2. Inventory Forecasting & Safety Stock Levels (Catalogue F)

```python
forecast_df = pd.read_csv('../outputs/inventory_forecast_reorder.csv')
print(forecast_df.head())

# Top 10 products by Reorder Point
top_reorder = forecast_df.nlargest(10, 'reorder_point')
plt.figure(figsize=(10, 4))
sns.barplot(data=top_reorder, x='product_id', y='reorder_point', palette='Blues_r')
plt.title('Top 10 Products by Reorder Point Threshold')
plt.xlabel('Product ID')
plt.ylabel('Reorder Point (Units)')
plt.xticks(rotation=45)
plt.show()
```
